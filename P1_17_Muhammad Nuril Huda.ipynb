Langkah 1

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

Langkah 2

```python
urls = [
    "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
    "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
    "https://external-preview.redd.it/hJNL3HXwmJXzzIdyUKdsVQerrBIe0KbO7d26zEXWidE.jpg?auto=webp&s=4a8bf4911d40dddb64ebc30505e704003297f414"
]

for url in urls:
  image = io.imread(url)                        #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  #convert color to RGB
  final_frame = cv.hconcat((image, image_2))      #concatenate image
  cv2_imshow(final_frame)                         #show image
  print('\n')
```

Langkah 3

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

Langkah 4

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

# **TUGAS PRAKTIKUM**

Buat garis vertikal dan garis menyilang diagonal pada image keluaran

```python
#Garis Vertikal
url = "https://external-preview.redd.it/hJNL3HXwmJXzzIdyUKdsVQerrBIe0KbO7d26zEXWidE.jpg?auto=webp&s=4a8bf4911d40dddb64ebc30505e704003297f414"

image= io.imread(url)

tinggi, lebar = image.shape[:2]

for x in range(tinggi):
    image[x, int(lebar/2)] = [255,255,255]

cv2_imshow(image)

print('\n')

#Garis Menyilang Diagonal
url = "https://external-preview.redd.it/hJNL3HXwmJXzzIdyUKdsVQerrBIe0KbO7d26zEXWidE.jpg?auto=webp&s=4a8bf4911d40dddb64ebc30505e704003297f414"

image = io.imread(url)

tinggi, lebar = image.shape[:2]

for y in range(tinggi):
    x1 = int(y * lebar / tinggi)
    x2 = lebar - x1 - 1

    image[y, x1] = [255,255,255]
    image[y, x2] = [255,255,255]

cv2_imshow(image)
```

Buat garis horisontal berwarna putih dibagian tengah gambar dengan panjang tertentu

```python
url = "https://danq.me/wp-content/uploads/2017/04/kitten-980-420.jpg"

image= io.imread(url)

tinggi, lebar = image.shape[:2]

for y in range(100, 400):
    image[int(tinggi/2), y] = [255,255,255]

cv2_imshow(image)
```

Buat kotak menggunakan kumpulan pixel warna putih di sembarang tempat dalam gambar

```python
url = "https://danq.me/wp-content/uploads/2017/04/kitten-980-420.jpg"

image= io.imread(url)

image[80:180, 80:180] = [255,255,255]

cv2_imshow(image)
```
